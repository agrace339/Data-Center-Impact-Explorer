```python
# x axis is region
# from 1 - 4, Northeast, Midwest, South, West so we have imbalances between scores and regions
import numpy as np
import matplotlib.pyplot as plt

cols = ['CURRESP_A','HLTHCOND_A', 'MEDRXTRT_A', 'COPDEV_A', 'CANEV_A', 'ASEV_A', 'DEPEV_A', 'ANXEV_A']

regions = sorted(dfnhis_refined['REGION'].unique())

fig, axes = plt.subplots(2, 4, figsize=(14, 9), sharex=True)
axes = axes.flatten()

for i, col in enumerate(cols):
    # count of 1s and 2s per region
    counts = dfnhis_refined.groupby('REGION')[col].value_counts().unstack(fill_value=0)
    counts = counts.reindex(regions)  # keeps region order consistent

    ax = axes[i]
    ax.bar(counts.index, counts[1], label='1', color='#4C72B0')
    ax.bar(counts.index, counts[2], bottom=counts[1], label='2', color='#DD8452') # I know for a fact I can see these colors

    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)
    ax.set_xticks([1, 2, 3, 4])
    ax.set_xticklabels(['Northeast', 'Midwest', 'South', 'West'])


axes[0].legend()
fig.suptitle('Response Counts (1 ("Yes") vs 2 ("No")) by Region', fontsize=14)
fig.axes[0].set_title("Respiratory allergies")
fig.axes[1].set_title("Weak Immune System (health cond.)")
fig.axes[2].set_title("Weak Immune System (medications)")
fig.axes[3].set_title("Ever had COPD")
fig.axes[4].set_title("Ever had cancer")
fig.axes[5].set_title("Ever had asthma")
fig.axes[6].set_title("Ever had depression")
fig.axes[7].set_title("Ever had anxiety")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

for i, col in enumerate(cols):

  contingency_table = pd.crosstab(dfnhis_refined['REGION'], dfnhis_refined[col])
  stat, pval, dof, expected = chi2_contingency(contingency_table)
  print(f"Chi-squared test for {col} and REGION: p-value = {pval:.4f}")

```

```python
from scipy.stats import chi2_contingency
import numpy as np

def cramers_v(confusion_matrix):
  chi2, p, dof, expected = chi2_contingency(confusion_matrix)
  n = confusion_matrix.sum().sum()
  phi2 = chi2 / n
  r, k = confusion_matrix.shape
  v = np.sqrt(phi2 / min(k - 1, r - 1))
  return chi2, p, v

for col in cols:
  table = pd.crosstab(dfnhis_refined['REGION'], dfnhis_refined[col])
  chi2, p, v = cramers_v(table)
  print(f"{col}: chi2={chi2:.2f}, p={p:.4g}, Cramér's V={v:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

cols = ['CURRESP_A', 'HLTHCOND_A', 'COPDEV_A', 'DEPEV_A', 'ANXEV_A']
col_labels = [
    "Respiratory\nallergies",
    "Weak immune\nsystem (health cond.)",
    "Ever had\nCOPD",
    "Ever had\ndepression",
    "Ever had\nanxiety",
]

region_names = {1: 'Northeast', 2: 'Midwest', 3: 'South', 4: 'West'}
regions = sorted(dfnhis_refined['REGION'].unique())

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
axes = axes.flatten()

x = np.arange(len(cols))

for i, region in enumerate(regions):
    ax = axes[i]
    sub = dfnhis_refined[dfnhis_refined['REGION'] == region]

    # captures the 1 and 2
    counts = sub[cols].apply(lambda s: s.value_counts()).reindex([1, 2], fill_value=0)

    # converting from raw counts to perecents
    pct = counts.div(counts.sum(axis=0), axis=1) * 100

    ax.bar(x, pct.loc[1], label='Yes', color='#4C72B0')
    ax.bar(x, pct.loc[2], bottom=pct.loc[1], label='No', color='#DD8452')

    ax.set_title(region_names[region])
    ax.set_xticks(x)
    ax.set_xticklabels(col_labels, rotation=45, ha='right')
    ax.set_ylabel('% of respondents')
    ax.set_ylim(0, 100)

axes[0].legend()
fig.suptitle('Percent of Response Rates (Yes vs No) by Health Variable, per Region', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
cols = ['CURRESP_A','HLTHCOND_A', 'MEDRXTRT_A', 'COPDEV_A', 'CANEV_A', 'ASEV_A', 'DEPEV_A', 'ANXEV_A']
col_labels = [
    "Respiratory\nallergies",
    "Weak immune\nsystem (health cond.)",
    "Weak immune\nsystem (medication)",
    "Ever had\nCOPD",
    "Ever had\ncancer",
    "Ever had\nasthma",
    "Ever had\ndepression",
    "Ever had\nanxiety",
]

pct_yes = pd.DataFrame(index=regions, columns=cols, dtype=float)
for region in regions:
    sub = dfnhis_refined[dfnhis_refined['REGION'] == region]
    counts = sub[cols].apply(lambda s: s.value_counts()).reindex([1, 2], fill_value=0)
    pct_yes.loc[region] = (counts.loc[1] / counts.sum(axis=0)) * 100


fig, ax = plt.subplots(figsize=(8, 5))
im = ax.imshow(pct_yes.values, cmap='Blues', vmin=0, vmax=100, aspect='auto')

ax.set_xticks(np.arange(len(cols)))
ax.set_xticklabels(col_labels, rotation=45, ha='right')
ax.set_yticks(np.arange(len(regions)))
ax.set_yticklabels([region_names[r] for r in regions])

for i in range(len(regions)):
    for j in range(len(cols)):
        ax.text(j, i, f'{pct_yes.values[i, j]:.0f}%', ha='center', va='center',
                color='black' if pct_yes.values[i, j] < 50 else 'white')

fig.colorbar(im, ax=ax, label='% responding "Yes"')
ax.set_title('Response Rate ("Yes") by Region and Health Variable')
plt.tight_layout()
plt.show()
```
